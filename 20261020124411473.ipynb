#**7) Veb sahifadan oʻqish**
1. Pandas yordamida veb sahifadagi jadvalni yuklab oling.
2. Ma’lumotlarni dataframe ko‘rinishida chiqaring.
3. Agar sahifada bir nechta jadval bo‘lsa, eng muhimini tanlang.
4. Ma’lumotni tozalang (bo‘sh qatorlarni olib tashlang, kerakli ustunlarni tanlang).
5. Natijani CSV faylga saqlang.

***Eslatma: manba sifatida oʻzingiz istagan veb sahifani tanlang. Bu vazifada mustaqil izlanishni oʻrganasiz!***

````python
import pandas as pd
import requests

headers = {
    'User-Agent': (
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML,'
        ' like Gecko) Chrome/122.0.0.0 Safari/537.36'
    )
}

url = 'https://en.wikipedia.org/wiki/London'

response = requests.get(url, headers=headers)#sahifani yuklab olaman

tables = pd.read_html(response.text)

df_web = tables[0]
df_web=df_web.dropna()#ma'lumotlarni tozalash
df_web.to_csv("london.csv", index=False)
df_web
````

* `chinook.db` maʼlumotlar omboriga `users` jadvali yarating. Foydalanuvchilarning har biri uchun berilgan yosh qiymatlaridan foydalanib, birth_date ustunini to‘ldirish strategiyasini ishlab chiqing va amalda qo‘llang. Jadval maydonlari (ustunlar):
