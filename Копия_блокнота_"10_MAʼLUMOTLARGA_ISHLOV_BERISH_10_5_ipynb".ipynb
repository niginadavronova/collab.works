#1-topshiriq


1.    Quyidagi link orqali maʼlumotlarni Pandas dataframeʼiga yuklab oling.
    
  *   https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/data_date.csv



2.   Yuklab olingan maʼlumotlarni `aqi.csv` fayliga saqlang.

3. Saqlab olingan `aqi.csv` faylining oxirgi 5 ta qatorini oʻqing.

Eslatma: fayl nomini `aqi.csv` deb saqlaganimizning sababi AQI – Air Quality Index, yaʼni havo sifati indeksi deganidir. Bizning maʼlumotlarimizda davlatlarning havo sifati haqidagi maʼlumotlar saqlangan!

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/data_date.csv")
df.to_csv("aqi.csv", index=False)
df.tail()
```

#2-topshiriq
1. Saqlab olingan `aqi.csv` faylidagi maʼlumotlarning faqat 2024-yilga tegishlisini excel faylga `aqi_2024.xlsx` koʻrinishida saqlab oling.
2. Saqlab olingan `aqi_2024.xlsx` faylidan Oʻzbekistonga tegishli maʼlumotni (qatorlarni) oʻqing.

```python
#javobni shu yerda yozing.
df=pd.read_csv("aqi.csv")
df_2024=df[df["Date"].str.startswith("2024")]
df_2024.to_excel("aqi_2024.xlsx", index=False)

aqi_excel=pd.read_excel("aqi_2024.xlsx")
uzb_2024=aqi_excel[aqi_excel["Country"]=="Uzbekistan"]
uzb_2024
```

#3-topshiriq
1. `aqi.csv` faylidagi maʼlumotlarni oʻrganib chiqsangiz, bu faylda 2022–2025-yillar oraligʻidagi havo sifati haqidagi maʼlumotlarni koʻrishingiz mumkin. Hozir sizga vazifa:
  * `aqi_2022.csv`, `aqi_2023.csv`, `aqi_2024.csv`, `aqi_2025.csv`  fayllariga `aqi.csv` faylidagi maʼlumotlarning mos davrlarini saqlang. Misol uchun, `aqi_2022` fayli 2022-yildagi havo sifati haqidagi maʼlumotlarni oʻzida saqlaydi.
  

```python
#javobni shu yerda yozing.
df=pd.read_csv("aqi.csv")
df_2022=df[df["Date"].str.startswith("2022")].to_csv("aqi_2022.csv", index=False)
df_2023=df[df["Date"].str.startswith("2023")].to_csv("aqi_2023.csv", index=False)
df_2024=df[df["Date"].str.startswith("2024")].to_csv("aqi_2024.csv", index=False)
df_2025=df[df["Date"].str.startswith("2025")].to_csv("aqi_2025.csv", index=False)
```

#4-topshiriq
1. Saqlab olingan `aqi_2022.csv`, `aqi_2023.csv`, `aqi_2024.csv`, `aqi_2025.csv`  fayllarini HDF5 formatida saqlang. Fayl nomini `aqi.h5` deb nomlang. Kalit (key) sifatida fayllarning yillarini bering. Misol uchun, key=2025.

```python
hdfdate = pd.HDFStore("aqi.h5")

hdfdate.put("2022", pd.read_csv("aqi_2022.csv"))
hdfdate.put("2023", pd.read_csv("aqi_2023.csv"))
hdfdate.put("2024", pd.read_csv("aqi_2024.csv"))
hdfdate.put("2025", pd.read_csv("aqi_2025.csv"))

hdfdate.close()
```

#5-topshiriq
1. `aqi.h5` faylidan maʼlumotlarni oʻqing va 2025-yilga tegishli maʼlumotlarni yangi dataframeʼga saqlang.
2. `aqi.h5` faylida 2025-yilning qaysi oylari uchun maʼlumotlar bor? Shu oylar roʻyxatini ekranga chiqaring!

```python
dfr_2025=pd.read_hdf("aqi.h5", key="2025")
oylar=dfr_2025["Date"].unique()
oylar
```

Vazifa: aqi.h5 faylidan istalgan yil (masalan, 2024) ma'lumotlarini o'qib oling va oylar bo'yicha havo sifati (AQI Value) ko'rsatkichining o'rtacha qiymatini (mean) hisoblang.

```python
df_24=pd.read_hdf("aqi.h5", key="2024")
orta=df_24["AQI Value"].mean()
orta
```

Qiymatlar bo'yicha saralash (sort_values)
Vazifa: 2025-yilga tegishli ma'lumotlarda havo sifati (AQI Value) eng yomon (eng yuqori) bo'lgan kunlarni aniqlash uchun jadvalni AQI Value bo'yicha kamayish tartibida (ascending=False) saralang va eng yuqori 5 ta qatorni chiqaring.

```python
df_25=pd.read_hdf("aqi.h5", key="2025")
bad_day=df_25.sort_values(by="AQI Value", ascending=False)
bad_day.head()
```
